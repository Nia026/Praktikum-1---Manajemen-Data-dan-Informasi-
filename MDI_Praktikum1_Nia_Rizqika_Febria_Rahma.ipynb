```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print('Seluruh library berhasil diimpor.')
```

```python
df = sns.load_dataset('flights')
df.head()
```

```python
print("Informasi struktur dataset")
df.info()
print("---------------------")
print("Tipe data setiap kolom")
print(df.dtypes)
print("---------------------")
print("Jumlah baris dan kolom")
print('Jumlah baris:', df.shape[0])
print('Jumlah kolom:', df.shape[1])
```

```python
penumpang_tahunan = df.groupby('year')['passengers'].sum().reset_index()
print('Jumlah Penumpang Setiap Tahun')
print(penumpang_tahunan)
```

```python
rata_rata_penumpang = df.groupby('year')['passengers'].mean().round(0).reset_index()
print('Rata-Rata Penumpang Setiap Tahun')
print(rata_rata_penumpang)
```

```python
idx_tertinggi = penumpang_tahunan['passengers'].idxmax()
tahun_tertinggi = penumpang_tahunan.loc[idx_tertinggi, 'year']
penumpang_tertinggi = penumpang_tahunan.loc[idx_tertinggi, 'passengers']

print(f'Tahun dengan penumpang terbanyak adalah tahun {tahun_tertinggi} dengan total {penumpang_tertinggi} penumpang')
```

```python
idx_rata_tertinggi = rata_rata_penumpang['passengers'].idxmax()
tahun_rata_tertinggi = rata_rata_penumpang.loc[idx_rata_tertinggi, 'year']
penumpang_rata_tertinggi = rata_rata_penumpang.loc[idx_rata_tertinggi, 'passengers']

print(f'Tahun dengan rata-rata penumpang terbanyak adalah tahun {tahun_rata_tertinggi} dengan rata-rata {penumpang_rata_tertinggi} penumpang')
```

```python
penumpang_awal = penumpang_tahunan.iloc[0]["passengers"]
penumpang_akhir = penumpang_tahunan.iloc[-1]["passengers"]

tahun_awal = penumpang_tahunan.iloc[0]["year"]
tahun_akhir = penumpang_tahunan.iloc[-1]["year"]

pertumbuhan = ((penumpang_akhir - penumpang_awal) / penumpang_awal) * 100

print(f"Pertumbuhan jumlah penumpang dari tahun {tahun_awal} ke tahun {tahun_akhir}: {pertumbuhan:.2f}%")
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    penumpang_tahunan["year"],
    penumpang_tahunan["passengers"],
    marker="o"
)

plt.title("Tren Jumlah Penumpang per Tahun")
plt.xlabel("Tahun")
plt.ylabel("Jumlah Penumpang")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    rata_rata_penumpang["year"],
    rata_rata_penumpang["passengers"],
    marker="o"
)

plt.title("Tren Rata-Rata Jumlah Penumpang per Tahun")
plt.xlabel("Tahun")
plt.ylabel("Jumlah Penumpang")
plt.grid(True)
plt.show()
```
